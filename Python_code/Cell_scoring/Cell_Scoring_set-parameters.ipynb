* replacements (dict): Dictionary where keys are texts to replace, values are replacements.
* cell_types (tuple): Cell types to apply the replacements in (e.g., 'code', 'markdown').

# 1. User Input

```python
# ===============Set the paths for the input and output================
#  Path to the template Jupyter Notebook
input_path = '/mypath/Cell_Scoring_analysis_template.ipynb' # <-- Replace this with the path of your template notebook
# Path to the modified Jupyter Notebook
output_path = '/resultpath/my_Cell_Scoring_analysis.ipynb' # <-- Replace this with the path where you want to save the modified notebook

# ===============Set the names of the markers==========================
# Receptors
receptor1 = 'Receptor1'   # <-- Replace this with the name of your receptor1
receptor2 = 'Receptor2'   # <-- Replace this with the name of your receptor2
```

# 2. Import Packages

```python
import nbformat
import pandas as pd
import os
import xlrd
```

# 3. Function

```python
def name_your_markers(input_path, output_path, replacements, cell_types=("code", "markdown")):
    """
    Replaces multiple texts in a Jupyter notebook.

    Parameters:
        input_path (str): Path to the input .ipynb file.
        output_path (str): Path to save the modified notebook.
        replacements (dict): Dictionary where keys are texts to replace, values are replacements.
        cell_types (tuple): Cell types to apply the replacements in (e.g., 'code', 'markdown').
    """
    with open(input_path, "r", encoding="utf-8") as f:
        notebook = nbformat.read(f, as_version=4)

    for cell in notebook.cells:
        if cell.cell_type in cell_types:
            source = cell.source
            for old, new in replacements.items():
                source = source.replace(old, new)
            cell.source = source

    with open(output_path, "w", encoding="utf-8") as f:
        nbformat.write(notebook, f)

    print(f"Applied replacements and saved to '{output_path}'")
```

# 4. Replace text

```python
# Set replacements for text in the modified notebook
replacements = {
    # Receptors
    'Receptor1': receptor1,
    'Receptor2': receptor2,
}

# function
name_your_markers(input_path, output_path, replacements)
# test that replacements were made
print(replacements)
```
