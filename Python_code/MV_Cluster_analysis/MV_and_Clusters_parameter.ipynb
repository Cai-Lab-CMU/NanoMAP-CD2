## 2. Set the replacements for text in the target Jupyter Notebook

```python
replacements = {
    # Receptors
    'markerA': markerA,
    'markerB': markerB,
    # Channels
    'Ch=mkrA': channel_markerA,         # channel for markerA 
    'Ch=mkrB': channel_markerB,         # channel for markerB
    'Ch=membrane': channel_membrane,     # channel for membrane ()
    'Ch=DistTrans': channel_distanceTransformation,    # channel for Distance Transformation 
    'Ch=binary_mkrA': channel_binary_markerA,  # channel for binary markerA 
    'Ch=binary_mkrB': channel_binary_markerB,  # channel for binary markerB
    'Ch=binary_MV': channel_binary_MV,    # channel for binary MV
}
```

## 3. Import Packages

```python
import nbformat
import pandas as pd
import os
import xlrd
```

## 4. Function
* name_your_markers

```python
def name_your_markers(input_path, output_path, replacements, cell_types=("code", "markdown")):
    """
    Replaces multiple texts in a Jupyter notebook.

    Parameters:
        input_path (str): Path to the input .ipynb file.
        output_path (str): Path to save the modified notebook.
        replacements (dict): Dictionary where keys are texts to replace, values are replacements.
        cell_types (tuple): Cell types to apply the replacements in (e.g., 'code', 'markdown').
    """
    with open(input_path, "r", encoding="utf-8") as f:
        notebook = nbformat.read(f, as_version=4)

    for cell in notebook.cells:
        if cell.cell_type in cell_types:
            source = cell.source
            for old, new in replacements.items():
                source = source.replace(old, new)
            cell.source = source

    with open(output_path, "w", encoding="utf-8") as f:
        nbformat.write(notebook, f)

    print(f"Applied replacements and saved to '{output_path}'")
```

## 5. Apply the replacement to generate the MV_and_Cluster_analysis.ipynb for specific dataset

```python
name_your_markers(input_path, output_path, replacements)
print(replacements)
```
